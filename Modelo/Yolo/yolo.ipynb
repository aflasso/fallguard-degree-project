# Entrenamiento, Pruebas y Validacion del modelo de deteccion de personas (YOLO)

```python
from ultralytics import YOLO
import cv2
```

```python
model = YOLO("yolo11m.pt")

# Entrenar
model.train(
    data="Data/Le2i_YOLO/data.yaml",
    epochs=50,
    imgsz=640,
)
```

```python
final_model = YOLO("runs/detect/train/weights/best.pt")
```

```python
results = final_model.track("../Data/Le2i/Home/Videos/video (35).avi", classes=[0], tracker="bytetrack.yaml")

for result in results:
    cv2.imshow("Person",result.plot())

    key = cv2.waitKey(0)  # 0 = espera hasta que presiones una tecla
    if key & 0xFF == ord('q'):
        break

cv2.destroyAllWindows()
```
